Groq Model

```python
from langchain_groq import ChatGroq

groq_llm = ChatGroq(
    model="llama-3.1-8b-instant"
)

response = groq_llm.invoke("Explain LangChain in simple words.")
print(response.content)
```

Gemini Model

```python
from langchain_google_genai import ChatGoogleGenerativeAI

gemini_llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash"
)

response = gemini_llm.invoke("What is Generative AI?")
print(response.content)
```

Prompt Template

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
    "Explain {topic} for beginners in 5 lines."
)

formatted = prompt.invoke({"topic": "Vector Database"})
print(formatted.text)
```

Simple Chain

```python
chain = prompt | groq_llm

result = chain.invoke({"topic": "LangChain"})
print(result.content)
```

Summarization Chain

```python
summary_prompt = PromptTemplate.from_template(
    "Summarize this text in 3 lines:\n{text}"
)

summary_chain = summary_prompt | gemini_llm

result = summary_chain.invoke({
    "text": "LangChain is a framework used to build modular AI applications using language models, tools, memory, and workflows."
})

print(result.content)
```

Tool Example

Memory Example

```python
chat_history = []

chat_history.append(("user", "My name is Krushna"))
chat_history.append(("assistant", "Nice to meet you"))
chat_history.append(("user", "What is my name?"))

for msg in chat_history:
    print(msg)
```

Mini Q&A App

```python
while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        break

    reply = groq_llm.invoke(user_input)
    print("Bot:", reply.content)
```
